```python
#Factorial
n = int(input("Enter Number:"))
fact = 1
for i in range(1,n+1):
    fact = fact * i
print("Factorial is:",fact)
```

```python
#Reverse Number
n = int(input("Enter Number:"))
rev = 0
while n > 0:
    rem = n % 10
    rev = rev * 10 + rem
    n = n // 10
print("Reverse is:",rev)
```

```python
#Prime
n = int(input("Enter Number:"))
for i in range(2,n):
    if n % i == 0:
        print("Not Prime")
        break
else:
    print("Prime")
```
